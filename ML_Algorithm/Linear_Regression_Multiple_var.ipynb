```python
df_exe = pd.read_csv('hiring.csv')
```

```python
df_exe
```

```python
from word2number import w2n
```

```python
df_exe.experience = df_exe.experience.fillna("zero")
```

```python
df_exe
```

```python
df_exe.experience = df_exe.experience.apply(w2n.word_to_num)
```

```python
df_exe
```

```python
import math
median_test_score = math.floor(df_exe['test_score(out of 10)'].mean())
median_test_score
```

```python
df_exe['test_score(out of 10)'] = df_exe['test_score(out of 10)'].fillna(median_test_score)
df_exe
```

```python
reg = LinearRegression()
reg.fit(df_exe[['experience','test_score(out of 10)','interview_score(out of 10)']],df_exe['salary($)'])
```

```python
reg.predict([[2,9,6]])
```

```python
reg.predict([[12,10,10]])
```
